Aligns each utterance with its speaker and emotion label

```python
import pandas as pd
import os


feat_csv = "/home/liaojd/SenticCrystal/scripts/MELD/opensmile/meld_egemaps_raw.csv"
df_feats = pd.read_csv(feat_csv)

# extract ids from current dialogue column in csv
df_feats['Dialogue_ID'] = df_feats['filename'].str.extract(r'dia(\d+)').astype(int)
df_feats['Utterance_ID'] = df_feats['filename'].str.extract(r'utt(\d+)').astype(int)

meta_paths = {
    'train': '/home/liaojd/SenticCrystal/data/meld_7way_data/train_sent_emo.csv',
    'dev': '/home/liaojd/SenticCrystal/data/meld_7way_data/dev_sent_emo.csv',
    'test': '/home/liaojd/SenticCrystal/data/meld_7way_data/test_sent_emo.csv'
}

all_meta = []
for split, path in meta_paths.items():
    if os.path.exists(path):
        m = pd.read_csv(path)
        m['split'] = split # ensure split matches
        all_meta.append(m)

df_meta = pd.concat(all_meta)

# merge on Dialogue_ID, Utterance_ID, and splits (test/train/val)
df_final = pd.merge(
    df_feats, 
    df_meta[['Dialogue_ID', 'Utterance_ID', 'Speaker', 'Emotion', 'Sentiment', 'split']], 
    on=['Dialogue_ID', 'Utterance_ID', 'split'], 
    how='left'
)

df_final.to_csv(feat_csv, index=False)
print("Updated CSV with Speaker and Label information!")
```
